# EDA 

```python
plt.figure(figsize=(28,10))
ax=sns.histplot(df['channel_title'],bins=20,color='red',kde=True)
for bars in ax.containers:
    ax.bar_label(bars)
plt.title('channel_title Distribution')
plt.xlabel('channel_title')
plt.ylabel('Distributions')
plt.show()
```

```python
plt.figure(figsize=(15,5))
ax=sns.countplot(x=df['category'],palette='Set2')
for bars in ax.containers:
    ax.bar_label(bars)
plt.title('category Distribution')
plt.xlabel('category')
plt.ylabel('Distributions')
plt.show()
```

```python
x=['views','likes','comments','trending_duration','engagement_rate','like_rate','comment_rate']
def univariate_analysis(n):
    plt.figure(figsize=(20, 24))
    for i, col in enumerate(n, 1):
        plt.subplot(4,2,i)
        plt.title(f"Count of {col}")
        sns.boxplot(x=df[col])
    plt.show()

univariate_analysis(x)
```
